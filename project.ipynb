```python
import tensorflow as tf
import numpy as np
import random
import matplotlib.pyplot as plt
import os
import keras
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
```

```python

# credit: https://www.kaggle.com/code/reidnersantos/gender-classification-cnn-image-dataset

size = 128
batch_size = 16
def load_dataset(directory, validation_split=None, subset=None):
    return tf.keras.utils.image_dataset_from_directory(
        directory,
        validation_split=validation_split,
        subset=subset,
        seed=123,
        image_size=(size, size),
        batch_size= batch_size)
    
test_data = load_dataset("C:/python/machine_learning/project/test",)
train_data = load_dataset("C:/python/machine_learning/project/train",)
```

```python
class_names = train_data.class_names
train_size = len(train_data) * batch_size
test_size = len(test_data) * batch_size
```

```python
men = []
women = []

for batch, labels in train_data:
    for img, label in zip(batch,labels):
        if label == 0:
            men.append((img, label))
            #print(np.min(img), np.max(img))
            #plt.imshow(img.numpy())
            #plt.show()
        else:
            women.append((img, label))

#maybe do data augmentation: https://keras.io/examples/vision/image_classification_from_scratch/

model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(128, 128, 3)),
    # First Convolutional Block
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    # Second Convolutional Block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    # Flattening and Dense Layers
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    
    # Output Layer: 1 neuron for binary classification (0 or 1)
    layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=5
)
```
